```python
def feet_to_inch(feet):
    return feet * 12
import math

BOTTOM_TO_TOP_OBS_VERT = feet_to_inch(3.115)
OBS_EDGE_HORZ = feet_to_inch(2.5)
OBS_EDGE_VERT = feet_to_inch(2)

def get_azimuth(mid_to_top_edge):
    MID_TO_TOP_EDGE = mid_to_top_edge #inch
    # MID_TO_TOP_OBS_VERT = feet_to_inch(3.115)
    if MID_TO_TOP_EDGE < BOTTOM_TO_TOP_OBS_VERT:
        print("Error: Mid to top edge is less than mid to top of obstacle, it must be greater than or equal {}".format(BOTTOM_TO_TOP_OBS_VERT))
        return 0
    MID_TO_BOTTOM_OBS_HORZ = math.sqrt(MID_TO_TOP_EDGE**2 - BOTTOM_TO_TOP_OBS_VERT**2)
    theta = math.asin((OBS_EDGE_HORZ/2)/MID_TO_BOTTOM_OBS_HORZ)
    # theta is half of the azimuth covered and we need that as well to use directly in the calculation using dish direction
    Azimuth =(math.degrees(theta))/1.5
    return Azimuth

def get_aoe(mid_to_top_edge):
    if mid_to_top_edge < BOTTOM_TO_TOP_OBS_VERT:
        print("Error: Mid to top edge is less than mid to top of obstacle, it must be greater than or equal {}".format(BOTTOM_TO_TOP_OBS_VERT))
        return 0
    MID_TO_TOP_MID = math.sqrt(mid_to_top_edge**2 - (OBS_EDGE_HORZ/2)**2)
    mid_to_top_angle = math.degrees(math.asin((BOTTOM_TO_TOP_OBS_VERT)/mid_to_top_edge))
    MID_TO_BOTTOM_MID = math.sqrt(MID_TO_TOP_MID**2 - (BOTTOM_TO_TOP_OBS_VERT)**2)
    MID_TO_BOTTOM_OBS_VERT = BOTTOM_TO_TOP_OBS_VERT - OBS_EDGE_VERT
    mid_to_bottom_angle = math.degrees(math.atan((MID_TO_BOTTOM_OBS_VERT)/MID_TO_BOTTOM_MID))
    print('obstructed area is from {} to {} and diff {}'.format(mid_to_bottom_angle, mid_to_top_angle, mid_to_top_angle - mid_to_bottom_angle))
    return mid_to_top_angle, mid_to_bottom_angle

def get_obs_area(mid_to_top_edge):
    Azimuth = get_azimuth(mid_to_top_edge)
    mid_to_top_angle, mid_to_bottom_angle = get_aoe(mid_to_top_edge)
    return Azimuth, mid_to_top_angle, mid_to_bottom_angle
print(get_azimuth(42))
print(get_aoe(42))

for i in [44,43,42]:
    print(get_azimuth(i))
    print(85 - get_azimuth(i), 85 + get_azimuth(i), 2*get_azimuth(i))
    print(get_aoe(i))
# print(get_obs_area(44))
```
